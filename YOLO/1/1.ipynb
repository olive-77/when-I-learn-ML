```python
def fun(x):
    return 2 * x ** 5 + x ** 4 - 11 * x ** 3 + 7 * x ** 2 + 2 * x
```

```python
import numpy as np
import matplotlib.pyplot as plt
i=-0.5
X=[]
Y=[]
for j in range(20):
    X.append(i)
    Y.append(fun(i))
    i += 0.1
X=np.array(X)
Y=np.array(Y)
X=np.c_[X**5,X**4,X**3,X**2,X]
w=np.random.randn(5)  #X@w
lr=0.0025
history=[]  #记录损失变化
one = np.ones(shape=(1,20))  #用来求和
for i in range(300000):
    y=X@w
    gradient=((y-Y).T@X).T
    w-=gradient*lr
    loss = one @ (y-Y)**2/20
    history.append(loss)
w  #[2,1,-11,7,2]
#这里轮次加到1000000就可以拟合的很完美了，但是我电脑不太行，这里存太多了后面内核会崩掉
```

```python
plt.plot(range(1,300001),history)
plt.show()
```
