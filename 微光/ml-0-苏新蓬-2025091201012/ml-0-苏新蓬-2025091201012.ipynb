# 3. NumPy基础

np.array()：最基础的方法，可将 Python 列表转换为 NumPy 矩阵，支持多维  
np.zeros()：创建指定形状的全零矩阵，默认数据类型为 float  
np.ones()：创建指定形状的全一矩阵   
np.eye(n)：创建 n×n 的单位矩阵（对角线为 1）   
np.arange() + reshape()：先创建有序序列，再调整为指定行列数  
np.random：生成随机值矩阵，常用的有rand()（均匀分布）和randn()（正态分布，初始方差为 1 ）

## 张量运算与广播

广播的「维度对齐优先级」:  
1.先给维度少的张量补全维度（从右向左补 1），使两者维度数量相同。  
2.再对每个维度从右向左检查：若维度大小不相等，只有其中一个为 1 时，才会将「1」扩展为另一个的大小。   
PS：如果step2中两边既不相等且都不是1，那么不满足广播规则，会报错  

```python
a = torch.tensor([[1], [2], [3]])  # shape (3, 1)
b = torch.tensor([4, 5, 6])        # shape (3,)\
# 计算 c = a + b，并写出 c 的形状
c = a+b   #3*3
# 计算 d = a * b，并解释广播规则  (广播过程会对)
d = a*b  #3*3
```
