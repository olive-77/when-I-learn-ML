# 1.在x,y的分布区域内均匀取点(x,y)计算对应函数值

```python
#设置x,y的值

import numpy as np
#x= np.random.uniform(low=-2,high=2,size =(1,10))
#y= np.random.uniform(low=-2,high=2,size =(1,10))
x=np.array([-2,-1.5,-1,-0.5,0,0.5,1,1.5,2])
y=np.array([0,0.5,1,1.5,2,2.5,3,3.5,4]).reshape(9,1)
zero=np.zeros(shape=(9,9))
x=(x+zero).reshape(1,-1)
y=(y+zero).reshape(1,-1)
d=np.ones((1,81))
e=0.1*np.random.randn(1,81)
```

```python
input=np.r_[x**3,y**4,np.exp(y),d,e]
w=np.array([3,1,1,5,1]).reshape(1,5)
F=w@input
F
```

# 2.使用Matplotlib库绘制出函数

```python
import matplotlib.pyplot as plt
fig = plt.figure (figsize=(10,7))
ax=fig .add_subplot(111,projection='3d')
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel('z')
ax.scatter(x,y,F,c=F)
```

# 3.梯度下降拟合

```python
input.T
```

```python
w_ =np.random.randn(1,5)
lr=0.00005
history=[]
for i in range (1000000):
    F_= w_ @ input
    gradient =2*(F_-F)@input.T/81
    w_-=gradient * lr
    if i%2000==0:
        loss =((F -F_)**2).mean()
        print(loss)
        history.append(float(loss))
print("w=",w_)
```

# 4.调整学习率

当lr大于0.011时，就会震荡不收缩，loss在迭代过程中显示inf和nan
当lr小于0.005时，收敛会很慢，进而导致相同迭代轮数下，最后结果与预期不是很接近

# 5. 绘制损失函数图像

```python
x1=[]
x2=[]
for i,j in enumerate(history):
    x1.append((i+1)*200)
    x2.append(j)
```

```python
fig1=plt.figure(figsize=(6,3))
ax=fig1.add_subplot(111)
ax.plot(x1,x2)
ax.set_xlabel('irritation')
ax.set_ylabel('Loss')
plt.show()
```

如果你说的时这种损失函数随迭代次数的变化图，那么是都能画的  
如果你说的是那种损失函数与各个参数大小的图，那么一旦次数上升到三维以上，就画不了了。二维的还可以用登高图表示一下
